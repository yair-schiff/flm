# FLM vs VDM Tau / Error Explorer

This notebook is a shareable, non-Streamlit version of the tau/error comparison views that now live in `scripts/flm_vdm_schedule_explorer.py`.

It reuses the same plotting and math helpers as the focused companion app `scripts/flm_vdm_tau_error_explorer.py`, so the figures here should stay aligned with the merged Streamlit UI.


## Setup

Run this notebook from the repo root so that imports like `scripts.flm_vdm_tau_error_explorer` and paths under `notes/` resolve correctly. For the interactive combined app, launch `scripts/flm_vdm_schedule_explorer.py`.


````python
from pathlib import Path

from IPython.display import Markdown, display

from scripts.flm_vdm_tau_error_explorer import (
    CompareParams,
    build_derivative_figure,
    build_error_figure,
    build_sampling_comparison_figure,
    build_tau_figure,
    compute_comparison,
    load_defaults,
)

defaults = load_defaults()
defaults
````

## Parameters

Edit these values and rerun the cells below to regenerate the figures.


````python
VOCAB_SIZE = 30522
GAMMA_MIN = defaults["gamma_min"]
GAMMA_MAX = defaults["gamma_max"]
T_PLOT_MAX = defaults["t_plot_max"]
PLOT_POINTS = 2048
N_GH = 100
TAU_SAMPLE_POINTS = 16

params = CompareParams(
    vocab_size=VOCAB_SIZE,
    gamma_min=GAMMA_MIN,
    gamma_max=GAMMA_MAX,
    t_plot_max=T_PLOT_MAX,
    plot_points=PLOT_POINTS,
    n_gh=N_GH,
    tau_sample_points=TAU_SAMPLE_POINTS,
)
params
````

````python
data = compute_comparison(params)

summary = {
    "snr_range": (float(data["snr_min"]), float(data["snr_max"])),
    "flm_p_error_at_t_max": float(data["p_error_flm"][-1]),
    "vdm_p_error_at_snr_max": float(data["p_error_vdm"][-1]),
    "chance_error": 1.0 - 1.0 / VOCAB_SIZE,
}
summary
````
